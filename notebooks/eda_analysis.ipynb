# Day 4 - Daily Returns Analysis

```python
nav_history = nav_history.sort_values(['amfi_code', 'date'])

nav_history['daily_return'] = (
    nav_history.groupby('amfi_code')['nav']
    .pct_change()
)

nav_history.head()
```

```python
avg_returns = (
    nav_history.groupby('amfi_code')['daily_return']
    .mean()
    .reset_index()
)

avg_returns.columns = ['amfi_code', 'avg_daily_return']

avg_returns.head()
```

```python
top_funds = avg_returns.sort_values(
    by='avg_daily_return',
    ascending=False
)

top_funds.head(10)
```

```python
top_funds.to_csv(
    "../reports/top_funds_daily_return.csv",
    index=False
)

print("Saved Successfully")
```

```python
cagr_data = nav_history.groupby('amfi_code').agg(
    start_nav=('nav', 'first'),
    end_nav=('nav', 'last')
).reset_index()

years = 3

cagr_data['cagr'] = (
    (cagr_data['end_nav'] / cagr_data['start_nav']) ** (1 / years) - 1
)

cagr_data.head()
```

```python
top_cagr = cagr_data.sort_values(
    by='cagr',
    ascending=False
)

top_cagr.head(10)
```

```python
top_cagr.to_csv(
    "../reports/top_cagr_funds.csv",
    index=False
)

print("Saved Successfully")
```

```python
import numpy as np

risk_free_rate = 0.065

sharpe_data = nav_history.groupby('amfi_code')['daily_return'].agg(
    ['mean', 'std']
).reset_index()

sharpe_data['sharpe_ratio'] = (
    (sharpe_data['mean'] * 252 - risk_free_rate)
    /
    (sharpe_data['std'] * np.sqrt(252))
)

sharpe_data.head()
```

```python
sharpe_rank = sharpe_data.sort_values(
    by='sharpe_ratio',
    ascending=False
)

sharpe_rank.head(10)
```

```python
sharpe_rank.to_csv(
    "../reports/sharpe_ratio_ranking.csv",
    index=False
)

print("Sharpe Ratio Saved Successfully")
```

```python
nav_history['running_max'] = (
    nav_history.groupby('amfi_code')['nav']
    .cummax()
)

nav_history['drawdown'] = (
    nav_history['nav'] /
    nav_history['running_max']
) - 1

nav_history.head()
```

worst drawdown per fund

```python
mdd_data = nav_history.groupby('amfi_code')['drawdown'] \
    .min() \
    .reset_index()

mdd_data.columns = ['amfi_code', 'max_drawdown']

mdd_data.head()
```

save report

```python
mdd_data.to_csv(
    "../reports/max_drawdown.csv",
    index=False
)

print("Maximum Drawdown Saved Successfully")
```
